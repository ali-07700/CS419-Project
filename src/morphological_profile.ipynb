# Morphological profile

This notebook scores one measurement: how much the small-scale bright and dark roof structure inside a known building changed between a pre-event satellite image and a post-event image. The score uses eight fixed morphological residuals on luminance: white and black disk top-hats at radii 1, 2, and 4 pixels, and white and black area top-hats at 16 pixels. The white area top-hat is the image minus its area opening, and the black area top-hat is the area closing minus the image. The reported score is the mean of the absolute differences of those eight residuals inside the polygon. A higher score means the small-scale surface structure changed more.

The morphological profile follows Gonzalez and Woods, *Digital Image Processing*, and Soille, *Morphological Image Analysis*. Top-hat residuals isolate bright peaks (such as rubble and roof edges) and dark valleys (such as collapse holes) smaller than the chosen scale, while remaining invariant to constant shifts in the roof's overall brightness. The sizes stay fixed and are not tuned on Mexico City or Palu.

The notebook does not read the other four scores and does not choose weights. Mexico City and Palu are not opened. Damage labels are joined only after the score exists, so the figures and the partial table can be drawn. The score function itself takes paired residual vectors and returns a number.

## Histogram matching

Singh (1989) and Lu et al. (2004) both require the two dates to share an intensity scale before a difference, or a residual comparison, means a change of the surface. Histogram matching is the point operation that does that. Lab 1 builds a histogram by counting gray levels and a cumulative distribution by `cumsum`. Gonzalez and Woods turn those two cumulative distributions into a lookup table: each post-event gray level is replaced by the pre-event gray level with the same cumulative value. Pratt treats the same table as a point operation \(s = T(r)\).

The match is done independently on the three color channels, with 256 bins. Pixels that are very dark and locally smooth are left out of both histograms, so a large dark water or shadow region does not set the roof scale. The constants used here are provisional and were not chosen on Mexico City or Palu: luminance below 40, and a 5 by 5 local standard deviation below 8. The lookup table is still applied to every pixel. Local standard deviation is the sample spread inside a box window, the same kind of neighborhood statistic used when Lab 1 and the spatial-filtering lectures separate a flat region from a textured one.

```python
def luminance(image):
    return cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)


def local_std(gray, window=5):
    values = gray.astype(np.float64)
    mean = cv2.blur(values, (window, window))
    second = cv2.blur(values * values, (window, window))
    variance = np.maximum(second - mean * mean, 0.0)
    return np.sqrt(variance)


def usable_pixels(image):
    gray = luminance(image)
    dark = gray < 40
    smooth = local_std(gray) < 8
    return ~(dark & smooth)


def match_channel(source, reference, source_keep, reference_keep):
    source_hist = np.bincount(source[source_keep], minlength=256).astype(np.float64)
    reference_hist = np.bincount(reference[reference_keep], minlength=256).astype(np.float64)
    if source_hist.sum() == 0 or reference_hist.sum() == 0:
        return source.copy()
    source_cdf = np.cumsum(source_hist) / source_hist.sum()
    reference_cdf = np.cumsum(reference_hist) / reference_hist.sum()
    lookup = np.searchsorted(reference_cdf, source_cdf, side="left")
    lookup = np.clip(lookup, 0, 255).astype(np.uint8)
    return lookup[source]


def histogram_match(post, pre):
    source_keep = usable_pixels(post)
    reference_keep = usable_pixels(pre)
    matched = np.empty_like(post)
    for channel in range(3):
        matched[:, :, channel] = match_channel(
            post[:, :, channel],
            pre[:, :, channel],
            source_keep,
            reference_keep,
        )
    return matched
```

## Local alignment

xBD pairs are already registered, but a roof can still sit a few pixels away from its pre-event footprint because the satellite is off-nadir. The workflow allows a shift of at most 4 pixels, and only when the short side of the polygon is at least 12 pixels. Smaller roofs are left where they are.

The shift is the peak of normalized cross-correlation, which is the template match in Lab 3. For a template \(T\) and a patch \(P\), both with their means removed,

\[
\mathrm{NCC} = \frac{\sum (T-\bar T)(P-\bar P)}{\sqrt{\sum (T-\bar T)^2 \sum (P-\bar P)^2}}.
\]

Lab 3 writes that sum in a loop. Here the search window is only 9 by 9, so `cv2.matchTemplate` with `TM_CCOEFF_NORMED` evaluates the same coefficient. Szeliski's frequency-domain phase correlation is the whole-image reserve and is not used on this subset. A flat template has no peak, and the shift stays zero.

```python
def polygon_points(wkt):
    start = wkt.find("((")
    end = wkt.rfind("))")
    body = wkt[start + 2:end]
    points = []
    for piece in body.split(","):
        parts = piece.split()
        points.append([float(parts[0]), float(parts[1])])
    return np.asarray(points, dtype=np.float64)


def short_side(points):
    width = float(points[:, 0].max() - points[:, 0].min())
    height = float(points[:, 1].max() - points[:, 1].min())
    return min(width, height)


def best_shift(pre_gray, post_gray, points):
    if short_side(points) < 12:
        return 0, 0
    height, width = pre_gray.shape
    x0 = max(0, int(np.floor(points[:, 0].min())))
    y0 = max(0, int(np.floor(points[:, 1].min())))
    x1 = min(width, int(np.ceil(points[:, 0].max())) + 1)
    y1 = min(height, int(np.ceil(points[:, 1].max())) + 1)
    if x1 - x0 < 2 or y1 - y0 < 2:
        return 0, 0
    template = pre_gray[y0:y1, x0:x1]
    if float(template.std()) < 1.0:
        return 0, 0
    search_x0 = max(0, x0 - 4)
    search_y0 = max(0, y0 - 4)
    search_x1 = min(width, x1 + 4)
    search_y1 = min(height, y1 + 4)
    search = post_gray[search_y0:search_y1, search_x0:search_x1]
    if search.shape[0] < template.shape[0] or search.shape[1] < template.shape[1]:
        return 0, 0
    scores = cv2.matchTemplate(search, template, cv2.TM_CCOEFF_NORMED)
    if not np.isfinite(scores).any():
        return 0, 0
    scores = np.nan_to_num(scores, nan=-1.0)
    _, _, _, peak = cv2.minMaxLoc(scores)
    dx = int(np.clip(peak[0] - (x0 - search_x0), -4, 4))
    dy = int(np.clip(peak[1] - (y0 - search_y0), -4, 4))
    return dx, dy
```

## Polygon interior and the morphological profile

The polygon is filled on the image grid and then eroded by one pixel with a 3 by 3 square, keeping unreliable roof boundaries out of the score. If fewer than 40 pixels survive, the original polygon is used and the building is still scored.

Inside that mask the score compares eight fixed morphological residuals on luminance after the post-event image has been histogram-matched:

1. **White disk top-hats** at radii \(r \in \{1, 2, 4\}\) pixels: \(W_r(I) = I - (I \circ B_r)\), where \(B_r\) is the discrete Euclidean disk of radius \(r\). These capture bright details smaller than the disk, such as bright rubble and sharp roof ridges.
2. **Black disk top-hats** at radii \(r \in \{1, 2, 4\}\) pixels: \(K_r(I) = (I \bullet B_r) - I\). These capture dark details smaller than the disk, such as collapse holes and narrow shadows between debris.
3. **White area top-hat** at \(\lambda = 16\) pixels: \(W_{16}^a(I) = I - \gamma_{16}^a(I)\), where \(\gamma_{16}^a\) is the grayscale 4-connected area opening removing bright connected structures smaller than 16 pixels regardless of shape.
4. **Black area top-hat** at \(\lambda = 16\) pixels: \(K_{16}^a(I) = \phi_{16}^a(I) - I\), where \(\phi_{16}^a\) is the grayscale 4-connected area closing removing dark connected holes smaller than 16 pixels.

Let \(R_{\mathrm{pre}}^{(k)}(p)\) and \(R_{\mathrm{post}}^{(k)}(p + \Delta)\) denote the \(k\)-th residual (\(k = 1, \dots, 8\)) inside the interior mask \(M\) and at the aligned post-event coordinates. The morphological profile score is

\[
\mathrm{score} = \frac{1}{8|M|} \sum_{k=1}^{8} \sum_{p \in M} \left| R_{\mathrm{post}}^{(k)}(p + \Delta) - R_{\mathrm{pre}}^{(k)}(p) \right|.
\]

An intact roof whose small-scale ridges and texture remain unchanged stays near 0, even if a uniform illumination offset remains. A collapsed roof with new rubble, holes, or vanished roof details produces a larger residual difference. All sizes stay fixed at \(r \in \{1, 2, 4\}\) and \(\lambda = 16\).

```python
DISK_RADII = (1, 2, 4)
AREA_THRESHOLD = 16
RESIDUAL_NAMES = (
    "white_disk_r1",
    "black_disk_r1",
    "white_disk_r2",
    "black_disk_r2",
    "white_disk_r4",
    "black_disk_r4",
    "white_area_16",
    "black_area_16",
)


def polygon_mask(points, shape):
    mask = np.zeros(shape, dtype=np.uint8)
    contour = np.round(points).astype(np.int32)
    cv2.fillPoly(mask, [contour], 1)
    return mask


def interior_mask(points, shape):
    mask = polygon_mask(points, shape)
    eroded = cv2.erode(mask, np.ones((3, 3), np.uint8), iterations=1)
    if int(eroded.sum()) < 40:
        return mask
    return eroded


def disk_element(radius):
    grid = np.arange(-radius, radius + 1)
    yy, xx = np.meshgrid(grid, grid, indexing="ij")
    return (xx * xx + yy * yy <= radius * radius).astype(np.uint8)


DISK_ELEMENTS = {radius: disk_element(radius) for radius in DISK_RADII}


def disk_tophats(gray, radii=DISK_RADII):
    residuals = []
    for radius in radii:
        kernel = DISK_ELEMENTS[radius]
        white = cv2.morphologyEx(gray, cv2.MORPH_TOPHAT, kernel)
        black = cv2.morphologyEx(gray, cv2.MORPH_BLACKHAT, kernel)
        residuals.append(white)
        residuals.append(black)
    return residuals


def area_opening_4conn(gray, area_threshold=AREA_THRESHOLD, target_mask=None):
    if HAS_SKIMAGE:
        return _sk_area_opening(gray, area_threshold=area_threshold, connectivity=1).astype(np.uint8)
    opened = np.zeros_like(gray, dtype=np.uint8)
    active_target = None if target_mask is None else (target_mask & (gray > 0))
    if active_target is not None and not np.any(active_target):
        return opened
    levels = np.unique(gray)
    for level in levels[levels > 0][::-1]:
        binary = (gray >= level).astype(np.uint8)
        _, labels, stats, _ = cv2.connectedComponentsWithStats(binary, connectivity=4)
        valid = stats[:, cv2.CC_STAT_AREA] >= area_threshold
        valid[0] = False
        keep = valid[labels]
        opened[keep & (opened == 0)] = level
        if active_target is not None and np.all(opened[active_target] > 0):
            break
    return opened


def area_tophats(gray, area_threshold=AREA_THRESHOLD, target_mask=None):
    if HAS_SKIMAGE:
        opened = _sk_area_opening(gray, area_threshold=area_threshold, connectivity=1).astype(np.int16)
        closed = _sk_area_closing(gray, area_threshold=area_threshold, connectivity=1).astype(np.int16)
        gray_i16 = gray.astype(np.int16)
        white = (gray_i16 - opened).astype(np.uint8)
        black = (closed - gray_i16).astype(np.uint8)
        return [white, black]
    opened = area_opening_4conn(gray, area_threshold=area_threshold, target_mask=target_mask)
    inverted = 255 - gray
    opened_inv = area_opening_4conn(inverted, area_threshold=area_threshold, target_mask=target_mask)
    white = gray - opened
    black = inverted - opened_inv
    return [white, black]


def compute_residual_stack(gray):
    layers = disk_tophats(gray) + area_tophats(gray)
    return np.stack(layers, axis=0)


def _extract_area_samples(gray, rows, cols, area_threshold=AREA_THRESHOLD):
    height, width = gray.shape
    r0 = max(0, int(rows.min()) - area_threshold)
    r1 = min(height, int(rows.max()) + 1 + area_threshold)
    c0 = max(0, int(cols.min()) - area_threshold)
    c1 = min(width, int(cols.max()) + 1 + area_threshold)
    patch = gray[r0:r1, c0:c1]
    target = np.zeros(patch.shape, dtype=bool)
    target[rows - r0, cols - c0] = True
    white_area, black_area = area_tophats(patch, area_threshold=area_threshold, target_mask=target)
    return np.vstack([
        white_area[rows - r0, cols - c0],
        black_area[rows - r0, cols - c0],
    ])


def paired_residuals(pre_gray, post_gray, pre_cache, post_cache, mask, dx, dy):
    rows, cols = np.nonzero(mask)
    shifted_rows = rows + dy
    shifted_cols = cols + dx
    height, width = pre_gray.shape
    keep = (
        (shifted_rows >= 0)
        & (shifted_rows < height)
        & (shifted_cols >= 0)
        & (shifted_cols < width)
    )
    rows = rows[keep]
    cols = cols[keep]
    shifted_rows = shifted_rows[keep]
    shifted_cols = shifted_cols[keep]
    if rows.size == 0:
        empty = np.empty((8, 0), dtype=np.uint8)
        return empty, empty
    if HAS_SKIMAGE:
        return pre_cache[:, rows, cols], post_cache[:, shifted_rows, shifted_cols]
    pre_disk = pre_cache[:, rows, cols]
    post_disk = post_cache[:, shifted_rows, shifted_cols]
    pre_area = _extract_area_samples(pre_gray, rows, cols)
    post_area = _extract_area_samples(post_gray, shifted_rows, shifted_cols)
    return np.vstack([pre_disk, pre_area]), np.vstack([post_disk, post_area])


def morphological_profile(pre_residuals, post_residuals):
    diff = np.abs(post_residuals.astype(np.float64) - pre_residuals.astype(np.float64))
    return float(diff.mean())
```

## Scoring the subset

Each pair is histogram-matched once and its disk top-hat maps (plus area top-hat maps when `scikit-image` is available) are computed once. Each classified polygon is then shifted, eroded, and scored on its own. The CSV columns are `image_id`, `building_id`, `score`, and `n_pixels`. `n_pixels` is the number of interior luminance pixels that entered the profile. The label is stored in memory for the next sections and is not an argument of `morphological_profile`.

```python
def load_pair(image_id):
    pre = cv2.imread(str(IMAGE_DIR / f"{image_id}_pre_disaster.png"), cv2.IMREAD_COLOR)
    post = cv2.imread(str(IMAGE_DIR / f"{image_id}_post_disaster.png"), cv2.IMREAD_COLOR)
    if pre is None or post is None:
        raise FileNotFoundError(image_id)
    label_path = LABEL_DIR / f"{image_id}_post_disaster.json"
    label = json.loads(label_path.read_text())
    return pre, post, label["features"]["xy"]


rows = []
for image_id in SUBSET:
    pre, post, features = load_pair(image_id)
    matched = histogram_match(post, pre)
    pre_gray = luminance(pre)
    post_gray = luminance(matched)
    if HAS_SKIMAGE:
        pre_cache = compute_residual_stack(pre_gray)
        post_cache = compute_residual_stack(post_gray)
    else:
        pre_cache = np.stack(disk_tophats(pre_gray), axis=0)
        post_cache = np.stack(disk_tophats(post_gray), axis=0)
    for feature in features:
        subtype = feature["properties"]["subtype"]
        if subtype == "un-classified":
            continue
        points = polygon_points(feature["wkt"])
        if len(points) < 3:
            continue
        mask = interior_mask(points, pre_gray.shape)
        if int(mask.sum()) == 0:
            continue
        dx, dy = best_shift(pre_gray, post_gray, points)
        pre_res, post_res = paired_residuals(
            pre_gray, post_gray, pre_cache, post_cache, mask, dx, dy
        )
        n_pixels = int(pre_res.shape[1])
        if n_pixels == 0:
            continue
        rows.append({
            "image_id": image_id,
            "building_id": feature["properties"]["uid"],
            "score": morphological_profile(pre_res, post_res),
            "n_pixels": n_pixels,
            "pre_mean": float(pre_gray[mask > 0].mean()),
            "subtype": subtype,
            "dx": dx,
            "dy": dy,
            "points": points,
        })

csv_path = TABLE_DIR / "morphology_subset.csv"
with csv_path.open("w", newline="") as handle:
    writer = csv.DictWriter(handle, fieldnames=["image_id", "building_id", "score", "n_pixels"])
    writer.writeheader()
    for row in rows:
        writer.writerow({
            "image_id": row["image_id"],
            "building_id": row["building_id"],
            "score": f"{row['score']:.6f}",
            "n_pixels": row["n_pixels"],
        })

print(f"buildings {len(rows)}")
print(f"csv {csv_path.resolve()}")
```

## Figures

Each example row is one building. The post-event crop is taken at the shift found for that polygon, so the outline sits on the aligned roof. The third panel shows the pixel-wise mean absolute difference across the eight morphological residuals (scaled by 4 for visibility) after that shift, highlighting where small-scale bright rubble and dark holes changed inside the outline.

The destroyed figure uses two buildings from different practice events. The intact figure uses one undamaged roof with a low score and one undamaged roof with a high score, showing where an undamaged roof still triggers top-hat residual differences. Buildings used in the panels have a short side of at least 24 pixels and at least 250 interior pixels.

The charts summarize the same 13 pairs: the score by damage class, the no-damage and destroyed histograms, and the mean score on each practice event. Minor and major damage are scarce here, so those two boxes are descriptive only. Mexico City and Palu are not in these figures.

```python
def crop_box(points, shape, margin=48):
    height, width = shape
    x0 = max(0, int(np.floor(points[:, 0].min())) - margin)
    y0 = max(0, int(np.floor(points[:, 1].min())) - margin)
    x1 = min(width, int(np.ceil(points[:, 0].max())) + margin)
    y1 = min(height, int(np.ceil(points[:, 1].max())) + margin)
    return x0, y0, x1, y1


def take_crop(image, box, dx=0, dy=0):
    x0, y0, x1, y1 = box
    height, width = image.shape[:2]
    src_x0 = min(max(x0 + dx, 0), width - 1)
    src_y0 = min(max(y0 + dy, 0), height - 1)
    src_x1 = min(max(x1 + dx, src_x0 + 1), width)
    src_y1 = min(max(y1 + dy, src_y0 + 1), height)
    crop = image[src_y0:src_y1, src_x0:src_x1]
    canvas = np.zeros((y1 - y0, x1 - x0) + image.shape[2:], dtype=image.dtype)
    dest_x = src_x0 - (x0 + dx)
    dest_y = src_y0 - (y0 + dy)
    canvas[dest_y:dest_y + crop.shape[0], dest_x:dest_x + crop.shape[1]] = crop
    return canvas


def draw_outline(axis, points, origin):
    relative = points - np.array(origin, dtype=np.float64)
    closed = np.vstack([relative, relative[:1]])
    axis.plot(closed[:, 0], closed[:, 1], color="#ffd60a", linewidth=1.6)
    axis.set_axis_off()


def show_case(axis, image, points, origin, title):
    axis.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    draw_outline(axis, points, origin)
    axis.set_title(title, fontsize=10)


def event_name(image_id):
    return image_id.rsplit("_", 1)[0]


def crop_residual_diff(pre_gray, post_gray, box, dx, dy):
    pad = AREA_THRESHOLD
    x0, y0, x1, y1 = box
    padded_box = (x0 - pad, y0 - pad, x1 + pad, y1 + pad)
    pre_patch = take_crop(pre_gray, padded_box)
    post_patch = take_crop(post_gray, padded_box, dx, dy)
    pre_stack = compute_residual_stack(pre_patch).astype(np.float64)
    post_stack = compute_residual_stack(post_patch).astype(np.float64)
    mean_abs = np.mean(np.abs(post_stack - pre_stack), axis=0)
    cropped = mean_abs[pad:pad + (y1 - y0), pad:pad + (x1 - x0)]
    return np.clip(np.round(cropped * 4.0), 0, 255).astype(np.uint8)


def panel_cases(cases, path, suptitle):
    fig, axes = plt.subplots(len(cases), 3, figsize=(9.2, 3.15 * len(cases)))
    if len(cases) == 1:
        axes = np.array([axes])
    cache = {}
    for axis_row, case in zip(axes, cases):
        if case["image_id"] not in cache:
            pre, post, _ = load_pair(case["image_id"])
            matched = histogram_match(post, pre)
            cache[case["image_id"]] = (pre, matched, luminance(pre), luminance(matched))
        pre, matched, pre_gray, post_gray = cache[case["image_id"]]
        box = crop_box(case["points"], pre.shape[:2])
        pre_crop = take_crop(pre, box)
        post_crop = take_crop(matched, box, case["dx"], case["dy"])
        difference = crop_residual_diff(pre_gray, post_gray, box, case["dx"], case["dy"])
        difference_bgr = cv2.cvtColor(difference, cv2.COLOR_GRAY2BGR)
        origin = (box[0], box[1])
        place = event_name(case["image_id"]).replace("-", " ")
        kind = "Destroyed" if case["subtype"] == "destroyed" else "No damage"
        show_case(axis_row[0], pre_crop, case["points"], origin, f"{kind}  {place}  pre-event")
        show_case(
            axis_row[1],
            post_crop,
            case["points"],
            origin,
            f"post-event, aligned   score {case['score']:.2f}",
        )
        show_case(axis_row[2], difference_bgr, case["points"], origin, "mean |Δ residual| (x4)")
    fig.suptitle(suptitle, fontsize=11)
    fig.tight_layout()
    fig.savefig(path, dpi=160, bbox_inches="tight")
    plt.show()
    for case in cases:
        print(case["image_id"], case["building_id"], f"{case['score']:.3f}", case["n_pixels"])


def visible_cases(subtype):
    pool = []
    for row in rows:
        if row["subtype"] != subtype or not np.isfinite(row["score"]) or row["n_pixels"] < 250:
            continue
        points = row["points"]
        width = float(points[:, 0].max() - points[:, 0].min())
        height = float(points[:, 1].max() - points[:, 1].min())
        short = min(width, height)
        long = max(width, height)
        if short < 24 or long > 160 or long / short > 2.4:
            continue
        if points[:, 0].min() < 20 or points[:, 1].min() < 20:
            continue
        if points[:, 0].max() > 1004 or points[:, 1].max() > 1004:
            continue
        pool.append(row)
    return sorted(pool, key=lambda row: row["score"])


destroyed_pool = visible_cases("destroyed")
intact_pool = visible_cases("no-damage")
if destroyed_pool:
    destroyed_high = destroyed_pool[-1]
    destroyed_other = next(
        (row for row in reversed(destroyed_pool) if event_name(row["image_id"]) != event_name(destroyed_high["image_id"])),
        destroyed_pool[0],
    )
    panel_cases(
        [destroyed_high, destroyed_other],
        FIGURE_DIR / "examples_destroyed.png",
        "Destroyed buildings on the practice subset. Not the evaluation.",
    )
if intact_pool:
    bright_intact = [row for row in intact_pool if row["pre_mean"] >= 90]
    panel_cases(
        [intact_pool[0], (bright_intact or intact_pool)[-1]],
        FIGURE_DIR / "examples_intact.png",
        "Undamaged buildings: a low score and a high score. Not the evaluation.",
    )

class_order = ["no-damage", "minor-damage", "major-damage", "destroyed"]
class_scores = [
    [row["score"] for row in rows if row["subtype"] == name]
    for name in class_order
]
all_scores = np.array([row["score"] for row in rows], dtype=np.float64)
upper_box = max(float(np.percentile(all_scores, 98)) * 1.1, 1.0) if all_scores.size else 1.0

fig, axis = plt.subplots(figsize=(6.4, 4.2))
axis.boxplot(class_scores, showfliers=False)
axis.set_xticks(
    range(1, len(class_order) + 1),
    [f"{name}\n(n={len(values)})" for name, values in zip(class_order, class_scores)],
)
axis.set_ylabel("Morphological profile score")
axis.set_ylim(0, upper_box)
axis.set_title("Practice subset only. Weights are not fit. Mexico City and Palu are excluded.")
fig.tight_layout()
fig.savefig(FIGURE_DIR / "score_by_class.png", dpi=160, bbox_inches="tight")
plt.show()

hist_values = np.array(class_scores[0] + class_scores[3], dtype=np.float64)
hist_upper = max(float(np.percentile(hist_values, 99.5)), 1.0) if hist_values.size else 1.0
bins = np.linspace(0, hist_upper, 25)
fig, axis = plt.subplots(figsize=(6.4, 4.2))
axis.hist(class_scores[0], bins=bins, density=True, alpha=0.55, label=f"no-damage (n={len(class_scores[0])})")
axis.hist(class_scores[3], bins=bins, density=True, alpha=0.55, label=f"destroyed (n={len(class_scores[3])})")
axis.set_xlabel("Morphological profile score")
axis.set_ylabel("Density")
axis.set_xlim(0, hist_upper)
axis.legend()
axis.set_title("Practice subset only. Minor and major damage are omitted.")
fig.tight_layout()
fig.savefig(FIGURE_DIR / "score_histogram.png", dpi=160, bbox_inches="tight")
plt.show()

event_order = []
for image_id in SUBSET:
    event = event_name(image_id)
    if event not in event_order:
        event_order.append(event)
grouped = {}
for name in ("no-damage", "destroyed"):
    grouped[name] = []
    for event in event_order:
        values = np.array([
            row["score"]
            for row in rows
            if row["subtype"] == name and event_name(row["image_id"]) == event
        ])
        grouped[name].append(float(values.mean()) if values.size else np.nan)
mean_vals = [v for v in grouped["no-damage"] + grouped["destroyed"] if np.isfinite(v)]
event_upper = max(mean_vals) * 1.15 if mean_vals else 1.0
positions = np.arange(len(event_order))
width = 0.36
fig, axis = plt.subplots(figsize=(7.2, 4.2))
axis.bar(positions - width / 2, grouped["no-damage"], width, label="no-damage")
axis.bar(positions + width / 2, grouped["destroyed"], width, label="destroyed")
axis.set_xticks(positions, [event.replace("-", "\n") for event in event_order])
axis.set_ylim(0, event_upper)
axis.set_ylabel("Mean morphological profile score")
axis.legend()
axis.set_title("Practice subset only. Mexico City and Palu are excluded.")
fig.tight_layout()
fig.savefig(FIGURE_DIR / "score_by_event.png", dpi=160, bbox_inches="tight")
plt.show()
```

## Partial numbers

A single threshold on the morphological profile score turns it into a destroyed-versus-rest decision. The threshold is swept over the observed score range to find the value with the highest destroyed-versus-rest F1 among those that keep the undamaged false-positive rate at or below 0.25. A false positive here is an undamaged building whose score is at least the threshold. The same cap is the one used later, when the five scores are weighted. This cut is provisional. It was fit on these 13 pairs only, and it will not be reused on Mexico City or Palu.

Precision, recall, and F1 use destroyed as the positive class and every other classified building as the negative class. Minor damage is too rare in this subset to support its own score. The threshold figure shows F1 and the undamaged false-positive rate across the range, with this cut marked.

```python
def binary_counts(y_true, y_pred):
    true_pos = int(np.sum(y_true & y_pred))
    false_pos = int(np.sum(~y_true & y_pred))
    false_neg = int(np.sum(y_true & ~y_pred))
    return true_pos, false_pos, false_neg


def f1_from_counts(true_pos, false_pos, false_neg):
    precision = true_pos / (true_pos + false_pos) if true_pos + false_pos else 0.0
    recall = true_pos / (true_pos + false_neg) if true_pos + false_neg else 0.0
    if precision + recall == 0:
        return 0.0, precision, recall
    return 2 * precision * recall / (precision + recall), precision, recall


scores = np.array([row["score"] for row in rows], dtype=np.float64)
destroyed = np.array([row["subtype"] == "destroyed" for row in rows])
undamaged = np.array([row["subtype"] == "no-damage" for row in rows])
max_score = max(float(np.ceil(scores.max())) if scores.size else 1.0, 1.0)
thresholds = np.round(np.linspace(0.0, max_score + 0.05, 251), 3)
f1_curve = []
rate_curve = []
best = None
for cut in thresholds:
    predicted = scores >= cut
    undamaged_rate = float(predicted[undamaged].mean()) if undamaged.any() else 1.0
    true_pos, false_pos, false_neg = binary_counts(destroyed, predicted)
    f1_value, precision, recall = f1_from_counts(true_pos, false_pos, false_neg)
    f1_curve.append(f1_value)
    rate_curve.append(undamaged_rate)
    if undamaged_rate > 0.25:
        continue
    candidate = (f1_value, -undamaged_rate, cut, precision, recall, undamaged_rate, true_pos, false_pos, false_neg)
    if best is None or candidate > best:
        best = candidate

f1, _, threshold, precision, recall, undamaged_rate, true_pos, false_pos, false_neg = best

fig, axis = plt.subplots(figsize=(6.6, 4.2))
axis.plot(thresholds, f1_curve, label="destroyed-versus-rest F1")
axis.plot(thresholds, rate_curve, label="undamaged false-positive rate")
axis.axvline(threshold, color="0.35", linestyle="--", label=f"chosen cut {threshold:.3f}")
axis.axhline(0.25, color="0.55", linestyle=":", label="false-positive cap 0.25")
axis.set_xlim(0, thresholds[-1])
axis.set_ylim(0, 1)
axis.set_xlabel("Threshold on morphological profile")
axis.set_ylabel("Rate")
axis.legend(fontsize=8)
axis.set_title("Provisional cut on the practice subset. Not Mexico City or Palu.")
fig.tight_layout()
fig.savefig(FIGURE_DIR / "threshold_curve.png", dpi=160, bbox_inches="tight")
plt.show()

lines = ["class, n, mean, median"]
for name in class_order:
    values = scores[np.array([row["subtype"] == name for row in rows])]
    if values.size == 0:
        lines.append(f"{name}, 0, , ")
        continue
    lines.append(f"{name}, {values.size}, {values.mean():.4f}, {np.median(values):.4f}")
lines.append("event, no-damage n, no-damage mean, destroyed n, destroyed mean")
for event in event_order:
    no_damage_values = [
        row["score"] for row in rows
        if row["subtype"] == "no-damage" and event_name(row["image_id"]) == event
    ]
    destroyed_values = [
        row["score"] for row in rows
        if row["subtype"] == "destroyed" and event_name(row["image_id"]) == event
    ]
    no_mean = f"{np.mean(no_damage_values):.4f}" if no_damage_values else ""
    destroyed_mean = f"{np.mean(destroyed_values):.4f}" if destroyed_values else ""
    lines.append(
        f"{event}, {len(no_damage_values)}, {no_mean}, {len(destroyed_values)}, {destroyed_mean}"
    )
lines.append(
    f"threshold, {threshold:.3f}, destroyed_f1, {f1:.4f}, precision, {precision:.4f}, recall, {recall:.4f}"
)
lines.append(
    f"undamaged_false_positive_rate, {undamaged_rate:.4f}, tp, {true_pos}, fp, {false_pos}, fn, {false_neg}"
)
lines.append("subset is 13 practice pairs. This cut is not frozen and was not applied to Mexico City or Palu.")
metrics_path = TABLE_DIR / "morphology_metrics.txt"
metrics_path.write_text("\n".join(lines) + "\n")
print("\n".join(lines))
```
